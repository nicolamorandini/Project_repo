# Erste API-Abfrage: Winterthur nach Zürich HB
Klicke oben rechts auf **Select Kernel**, wähle **livingapp**, und führe die Code-Zelle mit **Shift+Enter** aus.


```python
import requests
from datetime import date, timedelta

heute = date.today()
tage_bis_dienstag = (1 - heute.weekday()) % 7 or 7
reisedatum = heute + timedelta(days=tage_bis_dienstag)

url = 'https://transport.opendata.ch/v1/connections'
parameter = {
    'from': 'Winterthur',
    'to': 'Zürich HB',
    'date': reisedatum.isoformat(),
    'time': '08:00',
    'limit': 4,
    'transportations[]': 'train',
}
antwort = requests.get(url, params=parameter, timeout=30)
antwort.raise_for_status()
verbindungen = antwort.json()['connections']
for v in verbindungen:
    print('Fahrzeit:', v['duration'], '| Abfahrt:', v['from']['departure'])
```
